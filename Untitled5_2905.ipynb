```python
# Inicialização das variáveis
opcao = 0
numero_a = 0
numero_b = 0

# O menu principal fica fora do loop para não se repetir toda hora
print("==== MENU ====")
print("")
print("Opção 1: Somar")
print("Opção 2: Subtração")
print("Opção 3: Multiplicação")
print("Opção 4: Divisão")
print("==============")
print("")

# Loop principal do programa
while opcao != 5:
    opcao = int(input("Digite a opção desejada ou 5 para sair!: "))

    if opcao == 1:
        numero_a = float(input("Digite o primeiro número: "))
        numero_b = float(input("Digite o segundo número: "))
        print(f"O total da soma de {numero_a} + {numero_b} é igual a {numero_a + numero_b}")
        print("")

    elif opcao == 2:
        numero_a = float(input("Digite o primeiro número: "))
        numero_b = float(input("Digite o segundo número: "))
        print(f"O total da subtração de {numero_a} - {numero_b} é igual a {numero_a - numero_b}")
        print("")

    elif opcao == 3:
        numero_a = float(input("Digite o primeiro número: "))
        numero_b = float(input("Digite o segundo número: "))
        print(f"O total da multiplicação de {numero_a} x {numero_b} é igual a {numero_a * numero_b}")
        print("")

    elif opcao == 4:
        numero_a = float(input("Digite o primeiro número: "))
        numero_b = float(input("Digite o segundo número: "))
        # Proteção para não dividir por zero
        if numero_b == 0:
            print("Erro: Não é possível dividir por zero!")
        else:
            print(f"O total da divisão de {numero_a} / {numero_b} é igual a {numero_a / numero_b}")
        print("")

    elif opcao == 5:
        print("Saindo do programa... Até mais!")

    else:
        print("Opção inválida! Tente novamente.")
        print("")
```
