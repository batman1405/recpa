```python
# Programa Par ou Ímpar
try:
    # Solicita um número inteiro ao usuário
    numero = int(input("Digite um número inteiro: "))

    # Se o resto da divisão por 2 for 0, o número é par
    if numero % 2 == 0:
        print(f"O número {numero} é PAR.")
    else:
        print(f"O número {numero} é ÍMPAR.")

except ValueError:
    print("Erro: Por favor, digite apenas números inteiros.")

```

```python
# Programa Maior de Dois Números
try:
    # Solicita os dois números ao usuário
    num1 = float(input("Digite o primeiro número: "))
    num2 = float(input("Digite o segundo número: "))

    # Lógica de comparação
    if num1 > num2:
        print(f"O primeiro número ({num1}) é o MAIOR.")
    elif num2 > num1:
        print(f"O segundo número ({num2}) é o MAIOR.")
    else:
        print("Os dois números são IGUAIS.")

except ValueError:
    print("Erro: Por favor, digite apenas valores numéricos.")
```

```python
# Programa de Aprovação de Aluno
try:
    # Solicita a nota ao usuário
    nota = float(input("Digite a nota do aluno (0 a 10): "))

    # Lógica de verificação
    if nota >= 7:
        print(f"Status: APROVADO (Nota: {nota})")
    elif 5 <= nota < 7:
        print(f"Status: RECUPERAÇÃO (Nota: {nota})")
    else:
        print(f"Status: REPROVADO (Nota: {nota})")

except ValueError:
    print("Erro: Por favor, digite um valor numérico válido (use ponto em vez de vírgula).")
```

```python
# Programa de Classificação de Idade
try:
    # Solicita a idade ao usuário
    idade = int(input("Digite a idade: "))

    # Lógica de classificação
    if idade < 0:
        print("Erro: Idade não pode ser negativa.")
    elif 0 <= idade <= 12:
        print(f"Idade: {idade} anos - Classificação: CRIANÇA")
    elif 13 <= idade <= 17:
        print(f"Idade: {idade} anos - Classificação: ADOLESCENTE")
    elif 18 <= idade <= 59:
        print(f"Idade: {idade} anos - Classificação: ADULTO")
    else:
        print(f"Idade: {idade} anos - Classificação: IDOSO")

except ValueError:
    print("Erro: Por favor, digite um número inteiro válido para a idade.")
```

```python
# Programa de Verificação de Senha
# Define a senha correta (você pode mudar o que está entre aspas)
SENHA_CORRETA = "ninalinda"

# Solicita a senha ao usuário
senha_digitada = input("Digite a senha de acesso: ")

# Lógica de verificação
if senha_digitada == SENHA_CORRETA:
    print("Acesso Permitido! Seja bem-vindo.")
else:
    print("Acesso Negado! Senha incorreta.")
```

```python
# Programa Positivo, Negativo ou Zero
try:
    # Solicita um número ao usuário
    numero = float(input("Digite um número: "))

    # Lógica de classificação
    if numero > 0:
        print(f"O número {numero} é POSITIVO.")
    elif numero < 0:
        print(f"O número {numero} é NEGATIVO.")
    else:
        print("O número é ZERO.")

except ValueError:
    print("Erro: Por favor, digite um valor numérico válido.")
```

```python
# Programa Calculadora Simples
try:
    # Solicita os números e a operação
    num1 = float(input("Digite o primeiro número: "))
    operacao = input("Digite a operação (+, -, *, /): ")
    num2 = float(input("Digite o segundo número: "))

    # Lógica de cálculo
    if operacao == "+":
        resultado = num1 + num2
        print(f"Resultado: {num1} + {num2} = {resultado}")
    elif operacao == "-":
        resultado = num1 - num2
        print(f"Resultado: {num1} - {num2} = {resultado}")
    elif operacao == "*":
        resultado = num1 * num2
        print(f"Resultado: {num1} * {num2} = {resultado}")
    elif operacao == "/":
        # Verificação para evitar divisão por zero
        if num2 != 0:
            resultado = num1 / num2
            print(f"Resultado: {num1} / {num2} = {resultado}")
        else:
            print("Erro: Não é possível dividir por zero!")
    else:
        print("Operação inválida! Use apenas +, -, * ou /.")

except ValueError:
    print("Erro: Por favor, digite valores numéricos válidos.")
```

```python
# Programa Dia da Semana
try:
    # Solicita um número ao usuário
    dia = int(input("Digite um número de 1 a 7: "))

    # Lógica de verificação
    if dia == 1:
        print("1 = DOMINGO")
    elif dia == 2:
        print("2 = SEGUNDA-FEIRA")
    elif dia == 3:
        print("3 = TERÇA-FEIRA")
    elif dia == 4:
        print("4 = QUARTA-FEIRA")
    elif dia == 5:
        print("5 = QUINTA-FEIRA")
    elif dia == 6:
        print("6 = SEXTA-FEIRA")
    elif dia == 7:
        print("7 = SÁBADO")
    else:
        print("Erro: Número inválido! Digite apenas de 1 a 7.")

except ValueError:
    print("Erro: Por favor, digite um número inteiro.")

```

```python
# Programa de Ano Bissexto
try:
    # Solicita o ano ao usuário
    ano = int(input("Digite um ano (ex: 2026): "))

    # Regra: Ser divisível por 4 E (não ser divisível por 100 OU ser divisível por 400)
    if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
        print(f"O ano {ano} é BISSEXTO!")
    else:
        print(f"O ano {ano} NÃO É BISSEXTO.")

except ValueError:
    print("Erro: Por favor, digite um ano válido (número inteiro).")
```

```python
# Programa de Cálculo de Desconto
try:
    valor_compra = float(input("Digite o valor total da compra (R$): "))

    if valor_compra >= 100:
        desconto = 0.10  # 10%
        mensagem = "Desconto de 10% aplicado!"
    elif 50 <= valor_compra < 100:
        desconto = 0.05  # 5%
        mensagem = "Desconto de 5% aplicado!"
    else:
        desconto = 0     # Sem desconto
        mensagem = "Sem desconto para este valor."

    # Cálculos
    valor_desconto = valor_compra * desconto
    valor_final = valor_compra - valor_desconto

    # Resultados
    print("-" * 30)
    print(f"Valor original: R$ {valor_compra:.2f}")
    print(f"Status: {mensagem}")
    print(f"Valor do desconto: R$ {valor_desconto:.2f}")
    print(f"Total a pagar: R$ {valor_final:.2f}")
    print("-" * 30)

except ValueError:
    print("Erro: Digite um valor numérico válido.")
```
