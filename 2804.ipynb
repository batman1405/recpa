```python
# Escrever um programa que o usuário digita um número de 1 a 10 e o progrma que imprimi a tabuada do número digitado.

numero = int(input("Digite um número de 1 a 10: "))

print(f"\nTabuada do {numero}:")

# Faz o cálculo de 1 até 10
for i in range(1, 11):
    resultado = numero * i
    print(f"{numero} x {i} = {resultado}")
```
