```python
# Escrever um programa que recebe 04 notas referente a cada bimestre ecalcula a média das notase imprimi a média do aluno
#e se ele fooi aprovado, recuperação ou reprovado.

# Regras

# Utilizar while
# Definir uma variavel para o bimestre e uma variável para note e uma variavel para o resultado
```

```python
bimestre = 1
soma_notas = 0

while bimestre <= 4:
    # Pergunta a nota e transforma em número decimal
    nota = float(input("Digite a nota: "))

    # Soma a nota digitada ao total
    soma_notas = soma_notas + nota

    # Avança para o próximo bimestre
    bimestre = bimestre + 1

# Calcula a média dividindo por 4
media = soma_notas / 4

# Descobre a situação do aluno
if media >= 7:
    resultado = "Aprovado"

if media >= 5 and media < 7:
    resultado = "Recuperação"

if media < 5:
    resultado = "Reprovado"

# Mostra o resultado final na tela
print("Média final do aluno:")
print(media)
print("Situação:")
print(resultado)
```
